# GRPO 强化学习训练 - Kaggle 使用指南

## 1. 上传数据集
1. 将 `grpo_in_sandbox` 文件夹压缩为 zip
2. Kaggle -> Add Data -> Upload Dataset
3. 上传后的路径通常为: `/kaggle/input/grpo-in-sandbox/`

## 2. 导入模块

```python
import sys
sys.path.append('/kaggle/input/grpo-in-sandbox/')

from grpo_in_sandbox import train, RLHFTrainingConfig, ProductManager, RewardModel
```

## 3. 配置训练参数

```python
config = RLHFTrainingConfig(
    model_name_or_path = "Qwen/Qwen2.5-0.5B-Instruct",  # 或使用 Kaggle 上的模型
    num_train_epochs = 3,
    learning_rate = 1e-5,
    max_seq_length = 2048,
    temperature = 0.7,
    output_dir = "./rl_output",
)
```

## 4. 开始训练

```python
results = train(config)
```
